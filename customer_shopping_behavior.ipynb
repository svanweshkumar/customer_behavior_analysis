```python
import pandas as pd

df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.head()
```

```python
print(df.isnull().sum())
```

```python
df.info()
```

```python
df.describe(include = 'all')# for including categorical data in the summary statistics
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df.groupby('Category')['Review Rating'].transform('median'))
# instead of filling the missing values with the overall median, we fill them with the median of the respective category. This way, we preserve the inherent differences between categories and provide a more accurate representation of the data.
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')# to make the column names more consistent and easier to work with. This is a common practice in data analysis to ensure that column names are standardized and do not contain spaces, which can cause issues when referencing them in code.
```

```python
df.columns
```

```python
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
# creating a column with categories of ages
labels = ['Young Adult', 'Adult', 'Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age', 'age_group']].head()
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head()
```

```python
df['purchase_frequency_days'].isnull().sum()
```

```python
df[['discount_applied', 'promo_code_used']].head()
# is there any column where discount_applied is true but promo_code_used is false? This would indicate that a discount was applied without using a promo code, which could be an interesting insight into the shopping behavior of customers.
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop(columns = ['promo_code_used'] )
```
